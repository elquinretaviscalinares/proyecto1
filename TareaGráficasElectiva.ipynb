```python
import pandas as pd

df = pd.read_excel("ventas Zona centro 5.xlsx", skiprows=1) #saltamos la fila de titulo
df.columns = ["Mes", "2025", "2026"]
df = df[df["Mes"] != "Total general"] #quitamos la fila de total
df
```

```python
x = np.arange(len(df["Mes"]))
ancho = 0.35

plt.bar(x - ancho/2, df["2025"], width=ancho, label="2025", color="steelblue")
plt.bar(x + ancho/2, df["2026"], width=ancho, label="2026", color="orange")
plt.xticks(x, df["Mes"], rotation=45)
plt.title("Ventas Centro 5: 2025 vs 2026")
plt.ylabel("Ventas ($)")
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

```python
plt.plot(df["Mes"], df["2025"], marker="o", label="2025")
plt.plot(df["Mes"], df["2026"], marker="o", label="2026")
plt.title("Tendencia de ventas mes a mes")
plt.xlabel("Mes")
plt.ylabel("Ventas ($)")
plt.xticks(rotation=45)
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```
